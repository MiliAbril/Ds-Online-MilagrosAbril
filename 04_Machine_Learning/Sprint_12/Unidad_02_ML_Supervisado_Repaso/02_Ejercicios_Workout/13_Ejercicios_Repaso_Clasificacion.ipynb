### Ejercicio 1: Entendiendo el problema de "Negocio"

En función de la descripción del problema de "negocio" y sin mirar todavía los datos, ¿qué tipo de problema de clasificación vamos a resolver? ¿Qué métricas creess más conveniente para medir el rendimiento del modelo a construir y, por tanto, para comparar posibles modelos entre sí?

Es un problema de clasificación binaria (desarrollará cáncer de mama vs no lo desarrollará). Según la descripción del “negocio”, el error más costoso es el falso positivo: identificar como candidata a una persona que no desarrollará la enfermedad y exponerla innecesariamente a un tratamiento caro y con efectos secundarios. Por ello, la métrica principal a priorizar debería estar orientada a minimizar falsos positivos, es decir, maximizar la precisión y/o la especificidad (baja tasa de falsos positivos). De forma complementaria, conviene vigilar el recall/sensibilidad para no perder demasiados casos reales, ajustando el umbral de decisión para lograr un equilibrio acorde al objetivo clínico–económico descrito.
